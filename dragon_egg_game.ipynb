## Dragon Egg Text Adventure Game

```python

# the game's map
paths = {
        "hometown": ["desert", "forest"],
        "desert": ["forest", "beekeeper", "hometown"],
        "beekeeper": ["desert", "forest"],
        "forest": ["tower", "hometown"],
        "tower": ["clearing"]
        }


# descriptions of the locations in the game's map
descriptions = {
"hometown": "You are in your hometown ...",
"desert": "It's getting hotter. You reach the desert.",
"forest": "A dark and dangerous forest awaits you.",
"beekeeper": "You visit the beekeeper. You hope you don't get stung.",
"tower": '''
You finally reach the mage's tower. 
There is only one direction you can go!
         '''
                }


# initialize gamer location
room = "hometown"
# initialize the honey flag
honey = False

# game start output message
print("""
Find the Dragon Egg
===================

Your quest starts.
""")

while room != "clearing":    #condition to end the game

    print(f"\n{descriptions[room]}") # tell gamer where they currently are
    
    # conditions for special locations in the game
    if room == "tower":
        print('''
The mage gives you the spell you need before you leave the tower.
              '''
             )
        
    if room == "beekeeper":        

        if not honey:
            print("You buy a pot of honey at the beekeeper.")
            honey = True
        else: # if the gamer already visited the beekeeper
            print("You already have a pot of honey!")

    # tell gamer where they can go next
    print(f"You can go to {', '.join(paths[room])}.")
    target = input("Where do you want to go next? ").lower()
    
    # check if gamer gave an invalid location at any point in the game
    # tell gamer what happened next based on their input
    if target in paths[room]:       
        if target == "forest":
            if honey:
                print('''
You leave the honeypot to the bear and carefully sneak through.
                      '''
                     )
                honey = False
                room = "forest"
            else:
                print("There is a BEAR in the forest!!! You run away.")
                #target = room
        else:
            room = target # updating your location
    else:
        print("You can't go there.")

# after exit from while loop when room=clearing
print("""
On a hidden clearing you discover the dragon egg.

Your quest has been successful!
      """
     )
```
